### load experimental data 


```python
### first load min max ex metabolite concentration (from Canelas) ## with a margin for HCO3 values 
# cf # ###to add some uncertainty in the extracellular met bounds and match co2 default bounds computed by Bastian we will add 66% to the max and subtract 66% to the min
df_bounds_ex=pd.read_csv('../4_results/a_fluxspace/df_bounds_ex.csv',index_col=0)
#and fix bounds 
for met_id in df_bounds_ex.index:
    met=tmodel.metabolites.get_by_id(met_id)
    met.lower_bound=Q_(df_bounds_ex.loc[met_id,'lb'],'mM')
    met.upper_bound=Q_(df_bounds_ex.loc[met_id,'ub'],'mM')
    print(met.id)

```

```python
xiang_data=pd.read_excel('../1_data/b_processed_files/yeast_data_xiang_metabolites.xlsx',sheet_name='Assembly for Joana',index_col=1,header=[0,1])
display(xiang_data.head())
xiang_data=xiang_data.iloc[:,:-2].copy()
xiang_data=xiang_data.drop(index='not in the model')
#when cells have a < in it, replace it with nothing
##we need a value not a upper bound : we'll fit the treshold identified by Xiang
#the std will be higher than the value : 0.01 mmol/L
xiang_data.replace('<', '', regex=True, inplace=True)
xiang_data.replace(np.nan,0.01,inplace=True)
#there are 5 metabolites in dx that are not in WT/tm6 , and 9 in WT/tm6 that are not in dx

TM6_df=pd.DataFrame(index=pd.MultiIndex.from_product([['TM6'],xiang_data.index.tolist()]),columns=['mean','sd'])
TM6_df['mean']=xiang_data[('TM6','average')].values.astype(float)
TM6_df['sd']=xiang_data[('TM6','stdev')].values.astype(float)
#same for WT
WT_df=pd.DataFrame(index=pd.MultiIndex.from_product([['WT'],xiang_data.index.tolist()]),columns=['mean','sd'])
WT_df['mean']=xiang_data[('WT','average')].values.astype(float)
WT_df['sd']=xiang_data[('WT','stdev')].values.astype(float)
#concatenate the two dataframes and metabolite data
metabolite_data=pd.concat([TM6_df,WT_df])
metabolite_data.head()
```

```python
#load regression data
reg_data = hl.excel_to_df('../1_data/b_processed_files//extended_regressiondata_Canelas_Xiang_yeast.xlsx')
reg_data.keys()

reg_data['dvmeas'] = reg_data['dvmeas'].reset_index()
reg_data['dvmeas'].columns = ['cond','rxn','mean_sd','Value']
reg_data['dvmeas'] = reg_data['dvmeas'].replace({'biomass':'biomass_EX'})
reg_data['dvmeas'] = reg_data['dvmeas'].pivot(index=['cond', 'rxn'], columns='mean_sd', values='Value').rename_axis(None, axis=1)

### then create usual df with data


metabolite_data.drop(['gtp','cdp','ctp','gdp', 'nad','nadh', 'nadp', 'nadph'],level=1, axis=0, inplace=True) # ignore the NAD(P)H data as these are defined as ratios 
volume_data = pd.DataFrame({'WT':{'c':0.95, 'm':0.05}, 'TM6':{'c':0.90, 'm':0.1}}).T #specify the volume fractions for each condition

flux_data = reg_data['dvmeas']
flux_data=flux_data.loc[['WT','TM6']]
flux_data.drop(['o2_EX'],level=1, axis=0, inplace=True) #ignore the o2 flux data # specific to here
# flux_data#.to_csv('../processed_files/exflux_data.csv')
conditions=flux_data.index.get_level_values(0).unique()
print(conditions)

# flux_data.to_csv(f'{SYNDATA_FOLDER}../b_processed_files/xiangyeast_exflux_data.csv')
```

```python
### we make the databounds conc dataframe to expand default metabolite bounds #
CI=5
data_bounds_conc_xi = thermo_flux.solver.gurobi.calc_conc_bounds(tmodel, conds=conditions,  metabolite_data = metabolite_data,extracellular_data = None, 
                                                                 volume_data= volume_data, CI = CI, conc_units='M')

data_bounds_df=pd.DataFrame(index=[m.id for m in tmodel.metabolites if m.id in data_bounds_conc_xi.index.get_level_values(1)])

for i,condi in enumerate(conditions):
    data_bounds_df['lb'+str(i)]=data_bounds_conc_xi.loc[condi,'lb']
    data_bounds_df['ub'+str(i)]=data_bounds_conc_xi.loc[condi,'ub']
data_bounds_df_inf = data_bounds_df.replace(0, np.inf)
data_bounds_df['lb']=data_bounds_df[data_bounds_df>=0].min(axis=1) ##non negative because sometimes with meas-CI*sd we get negative values
data_bounds_df['ub']=data_bounds_df.max(axis=1)

for i,row in data_bounds_df[['lb','ub']].iterrows():
    lb=row['lb']
    ub=row['ub']
    if lb>0:
        tmodel.metabolites.get_by_id(i).lower_bound=min(tmodel.metabolites.get_by_id(i).lower_bound,Q_(lb, 'M'))
    else : 
        print(i,'here')
    tmodel.metabolites.get_by_id(i).upper_bound=max(Q_(ub, 'mM'),tmodel.metabolites.get_by_id(i).upper_bound)
    #increase default bounds
    print(i,tmodel.metabolites.get_by_id(i).lower_bound,tmodel.metabolites.get_by_id(i).upper_bound)
# tmodel.metabolites.atp_c.lower_bound=Q_(0,'mM')
```
